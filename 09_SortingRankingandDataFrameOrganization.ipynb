## Sorting, Ranking and DataFrame Organization

### 1. SORTING VALUES — `sort_values()`

```python
# Ascending order (smallest to largest)
df.sort_values(by="Salary")

# Descending order (largest to smallest)
df.sort_values(by="Salary", ascending=False)

# Sort by multiple columns
df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)

# Save the sorted DataFrame
df = df.sort_values(by="Salary", ascending=False)

# Modify the original DataFrame directly
df.sort_values(by="Salary", inplace=True)
```

### 2. SORTING INDEX — `sort_index()`

```python
# Sort rows by index
df.sort_index()

# Descending index order
df.sort_index(ascending=False)

# Sort columns alphabetically
df.sort_index(axis=1)
```

- `axis=0`: sort rows by index (default).
- `axis=1`: sort columns by their labels.

### 3. RANKING — `rank()`

```python
# Rank lowest salary as 1
df["Rank"] = df["Salary"].rank()

# Rank highest salary as 1
df["Rank"] = df["Salary"].rank(ascending=False)

# Choose how ties are handled
df["Rank"] = df["Salary"].rank(method="average")
df["Rank"] = df["Salary"].rank(method="min")
df["Rank"] = df["Salary"].rank(method="max")
df["Rank"] = df["Salary"].rank(method="first")
df["Rank"] = df["Salary"].rank(method="dense")
```

**Ranking methods:**
- `average`: tied values receive the average rank (default).
- `min`: tied values receive the lowest rank in the tie.
- `max`: tied values receive the highest rank in the tie.
- `first`: ranks ties in their order of appearance.
- `dense`: tied values share a rank; the next distinct value gets the next consecutive rank.

### 4. TOP AND BOTTOM ROWS

```python
# Three highest salaries
df.nlargest(3, "Salary")

# Three lowest salaries
df.nsmallest(3, "Salary")

# Top three using multiple columns
df.nlargest(3, ["Salary", "Age"])
```

### 5. RESETTING THE INDEX

```python
# Reset index, retaining the old index as a column
df.reset_index()

# Reset index and discard old index
df.reset_index(drop=True)

# Save the reset index
df = df.reset_index(drop=True)
```

### 6. SETTING THE INDEX

```python
# Use Employee_ID as index
df.set_index("Employee_ID")

# Keep Employee_ID as a regular column too
df.set_index("Employee_ID", drop=False)

# Save the change
df = df.set_index("Employee_ID")
```

### 7. GROUPBY + RANKING

```python
# Rank salaries within each department
df["Department_Rank"] = (
    df.groupby("Department")["Salary"]
      .rank(ascending=False)
)

# Dense ranking within each department
df["Department_Rank"] = (
    df.groupby("Department")["Salary"]
      .rank(ascending=False, method="dense")
)

# Highest salary in each department
df.groupby("Department")["Salary"].max()

# Sort department alphabetically, salary highest first
df.sort_values(
    by=["Department", "Salary"],
    ascending=[True, False]
)
```

### 8. IMPORTANT DIFFERENCES

| Method | Purpose |
|---|---|
| `sort_values()` | Sort by column values |
| `sort_index()` | Sort by index labels |
| `rank()` | Assign ranks to values |
| `nlargest()` | Return rows with largest values |
| `nsmallest()` | Return rows with smallest values |
| `reset_index()` | Reset the index |
| `set_index()` | Set a column as the index |
| `groupby().rank()` | Rank values within groups |

### 9. KEY REMINDERS

- Sorting methods return a new DataFrame by default.
- Use assignment or `inplace=True` to retain sorting changes.
- `rank(ascending=False)` gives the largest value rank 1.
- `dense` ranking does not skip ranks after ties.
- `reset_index(drop=True)` discards old index labels.
- `groupby().rank()` calculates ranks separately for each group.
- Missing numeric values are generally excluded from ranking and receive `NaN` ranks by default.

>Ranking with rank()

````python
# Assign a rank to each value in a column.
# By default, the smallest value gets rank 1, and tied values receive the average of their ranks.
df["Salary_Rank"] = df["Salary"].rank()
df
````

````python
df["Salary_Rank"] = df["Salary"].rank(ascending=False) #now largest values will get rank 1
df
````

````python
#ranking methods
series = pd.Series([100, 100, 200, 300, 500, 400], index=["a", "b", "c", "d", "e", "f"])
series = series.rank(method="average") #default
series
````

````python
series = series.rank(method="min")
series
````

````python
series = series.rank(method="max", ascending=False)
series
````

Method	    ---            How ties are handled
- average	---            Average of the tied ranks (default)
- min	    ---            Give all tied values the lowest rank in the group
- max	    ---            Give all tied values the highest rank in the group
- first 	---            Assign ranks according to their order of appearance
- dense 	---            Tied values share a rank; the next distinct value gets the next consecutive rank
