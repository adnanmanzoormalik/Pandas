```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Adnan", "Asrar", "Saad", "Zoya", "Mehak", "Adnan", "Rohan", "Sara"],
    "Age": [26, 24, 22, np.nan, 23, 26, 29, 21],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT", "Finance", "HR"],
    "Salary": [50000, 45000, 35000, 60000, np.nan, 55000, 70000, 40000],
    "City": ["Srinagar", "Jammu", "Srinagar", "Delhi", "Jammu", np.nan, "Delhi", "Srinagar"]
})

df
```

```python
# Find employees in the IT department whose salary is at least 50,000. Display only Name, Salary, and City, and sort by Salary from highest to lowest.
df[["Name","Salary","City"]].query("Salary >= 50000").sort_values("Salary", ascending=False)
```
