```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Adnan", "Asrar", "Saad", "Zoya", "Akash", "Mehak"],
    "Age": [26, 24, 22, None, 26, 23],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"],
    "Salary": [50000, 45000, 35000, 60000, np.nan, 45000],
    "City": ["Srinagar", "Jammu", "Srinagar", "Delhi", np.nan, "Jammu"]
})

df
```

> SELECTING ROWS
using loc[] and iloc[] >>> most important selection tools in pandas

```python
# loc[] — Label-based selection --- we have default indexes here i.e 0,1,2,3,4
df.loc[0] #Returns the row whose index label is 0.
```

```python
df.loc[0, "Name"] #selecting a row and column
```

```python
df.loc[0:2] #multiple rows
```

```python
df.loc[0:2, ["Name","Age"]] #multiple rows and columns
#NOTE: in this 0 and 2 both are included
```

```python
df.loc[0:4:2] #slicing
```

```python
df.loc[[1,2,0]] #multiple non consecituve rows
```

```python
# iloc[] — Position-based selection
df.iloc[0] #returns ist row
```

```python
df.iloc[0,0] #returns first row and column
```

```python
df.iloc[0:2] #returns multiple rows 
#NOTE: 0 is inclusive but 2 is exclusive
```

```python
df.iloc[0:2, 0:3] #first 2 rows and first 3 columns
```

```python
df.iloc[[1,2,3],[2,1]] #non consecutive rows and columns
```

> Filter rows using comparison operators
Filtering means selecting only the rows that satisfy a condition.

```python
df[df["Salary"]>45000]
#in this df["Salary"]>45000 returns True and then outer df only keeps those rows which return True
```

```python
df[df["Age"]>25]
```

we can use ==   >=   <=   !=   <   > for comparison 
and also &  |  ~  (AND, OR, NOT)

```python
df[(df["Age"]>25) | (df["Salary"]>45000)]
# df[df["Age"] >= 23 & df["Salary"] >= 45000] # >>> incorrect
```

```python
df[~(df["Department"] == "IT")]
```

> Advanced filtering: isin(), between(), and query()

isin() is convenient for matching several possible values.

between() is convenient for numeric or comparable ranges.

query() provides a readable expression-based alternative for many filtering tasks.

Boolean indexing with & and | is the most general and flexible approach.

```python
#isin() — Match values from a list
df[df["Department"].isin(["IT","Finance"])]
```

```python
df[~df["Department"].isin(["IT","Finance"])] #returns depts which are not IT and Finance
```

```python
# between() — Filter within a range
df[df["Salary"].between(40000, 50000)] #40000 and 50000 are both inclusive
```

```python
# query() — Filter with an expression
df.query("Salary > 45000") #Find employees with salaries above ₹45,000:
```

```python
df.query("Department== 'IT' and Age > 25") #Find IT employees aged at least 25:
```

```python
#using a python variable in query()
min_amt = 40000
df[["Department", "Age", "Salary"]].query("Salary > @min_amt")
```

> Sort data using sort_values() and sort_index()

```python
# sort_values() — Sort by column values
df.sort_values("Salary") #by default sort_value() is ascending
```

```python
df.sort_values("Salary", ascending=False)
```

```python
df.sort_values(["Department","Salary"],ascending=[False, False])
```

```python
# By default, missing values are placed at the end. You can change that:
df.sort_values("Salary", ascending=False,na_position="first")
```

```python
# sort_index() — Sort by index labels
df.sort_index()
```

```python
df.sort_index(ascending=False)
```

```python
df.sort_index(axis=1)
# To sort columns by their column labels:
```
