The goal is to learn how to inspect a dataset, understand its structure, identify data types, summarize values, and detect missing data before performing any analysis.

1 Inspect rows: head(), tail(), sample()

2 Understand structure: shape, columns, index, info(), dtypes

3 Summarize data: describe(), value_counts(), nunique()

4 Detect missing values: isna(), notna()

5 Understand counting: size, shape, count()

```python
#practice dataset
#This dataset has six employee records, repeated names and departments, and some missing values (NaN). That lets us practise nearly every inspection method without creating separate datasets.

import pandas as pd

df = pd.DataFrame({
    "Name": ["Adnan", "Asrar", "Saad", "Zoya", "Adnan", "Mehak"],
    "Age": [26, 24, 22, None, 26, 23],
    "Department": ["IT", "HR", "IT", "Finance", "IT", "HR"],
    "Salary": [50000, 45000, 35000, 60000, None, 45000],
    "City": ["Srinagar", "Jammu", "Srinagar", "Delhi", None, "Jammu"]
})

df
```

```python
# info() — Dataset overview
# This is one of the most useful inspection methods. It shows:
# - Number of rows and columns.
# - Column names.
# - Non-null counts for each column.
# - Data types.
# - Memory usage.
df.info()
# Age is a float64 instead of an integer because the column contains a missing value (NaN), which commonly causes a standard integer column to be represented as floating point.
# Important: info() prints its summary and returns None. You normally run df.info() by itself rather than assigning its result to a variable.
```

```python
df.dtypes #Data types of columns
```

```python
#we can also do this
df["Salary"].dtype
```

```python
# By default, describe() summarizes numeric columns. It provides:
# Statistic	Meaning
# count	Number of non-missing values
# mean	Average
# std	Standard deviation
# min	Minimum
# 25%	First quartile
# 50%	Median
# 75%	Third quartile
# max	Maximum

df.describe()
```

```python
df.describe(include="all")
# summarizes non-numeric columns, where applicable. For categorical columns, you'll typically see:
# - count: Non-missing values.
# - unique: Number of distinct values.
# - top: Most frequent value.
# - freq: Frequency of the most frequent value.
```

```python
# Count occurrences: value_counts()
df["Department"].value_counts()
```

```python
#By default, value_counts() excludes missing values.
df["City"].value_counts(dropna=False)
```

```python
#Normalize to proportions
df["Department"].value_counts(normalize=True)
```

```python
#to percentage
df["Department"].value_counts(normalize=True)*100
```

```python
# Count distinct values: nunique()
df["City"].nunique()
```

```python
df["City"].nunique(dropna=False)
```

```python
#unique() vs nunique()
print(df["City"].unique()) #returns list of distinct cities
print(df["City"].nunique()) #return number of distinct cities
```

```python
# isna() — Identify missing values
df.isna() #this return True or False for every cell/value
```

```python
df.isna().sum() #return number of na values in each column
```

```python
df.isna().sum().sum() #returns total number of na in dataframe
```

```python
#notna() - Identify non-missing values
df.notna() 
```

```python
df.notna().sum()
```

```python
df[df.isna().any(axis=1)] #This returns rows containing at least one missing value.
```

```python
# Understand shape, size and count()
print(df.shape) #Returns the number of rows and columns as a tuple: (6, 5).
print()
print(df.size) #Returns the total number of cells, including missing cells: 30.
print()
print(df.count()) #Returns the number of non-missing values in each column.
```
