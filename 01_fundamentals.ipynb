> Pandas: is a Python library used to manipulate, analyze, clean, and organize structured data.
Think of it as a tool for working with data in tables, similar to Excel spreadsheets or SQL query results, but directly in Python.
Common applications include:
- Data cleaning and handling missing values.
- Analyzing sales, employee, and customer datasets.
- Reading and processing CSV and Excel files.
- Summarizing data and calculating statistics.
- Preparing datasets for visualization and machine learning.
- Performing data-analysis tasks in entry-level interviews.



These are the two fundamental Pandas data structures.

Series: A one-dimensional labeled array, usually representing a single column of data.
DataFrame: A two-dimensional labeled table containing rows and columns.

>Series
A Series contains values and an associated index(index labels).

```python
marks = pd.Series([10,20,30])
print(marks)
```

```python
marks = pd.Series([10,20,40,30], index=["a","b","c","d"])
print(marks)
```

```python
#CREATING FROM A DICTIONARY
marks = pd.Series({
        "Adnan":10,
        "Asrar":20,
        "Saad":30
    })
print(marks)
```

> DataFrame
A DataFrame is the primary object you'll use throughout Pandas.

```python
#Create from a dictionary of lists
#Each dictionary key becomes a column name, and each list supplies the values in that column.
students = {
    "Name":["Adnan","Asrar","Saad"],
    "Age":[25,30,3],
    "Marks":[10,20,30]
}
df = pd.DataFrame(students)
print(df)
```

```python
#creating from a list of dictionaries
students = [
    {"Name":"Adnan", "Age":20},
    {"Name":"Asrar", "Age":2},
    {"Name":"Saad", "Age":0}
]
df = pd.DataFrame(students)
df
```

```python
#Create form a numoy array
import numpy as np
import pandas as pd

arr = np.array([
    [1, 85],
    [2, 90],
    [3, 78]
])

df = pd.DataFrame(
    arr,
    columns=["Student_ID", "Marks"]
)

df
```

```python
#Create a df with custom index
df = pd.DataFrame({
    "Name":["Adnan", "Asrar"],
    "Age":[25,30]
}, index=["a","b"])
df
```

```python
# underssting rows, columns, indexes, column names
df = pd.DataFrame(
    {
        "Name": ["Aman", "Sara", "Rohan"],
        "Age": [21, 22, 20],
        "Marks": [85, 90, 78]
    },
    index=["S1", "S2", "S3"]
)
df
```

```python
df.index #Remember: an index is a label, not necessarily a row number. It can contain strings, dates, or other valid labels.
```

```python
#Suppose you have a file named students.csv in the same folder as your notebook:
df = pd.read_csv("students.csv")
df
```

```python
df = pd.read_csv("students.csv",nrows=2) #this ll read only first 2 rows
df
```

```python
df = pd.read_csv("students.csv", sep=";") #this ll find the ; in csv and use that as delimiter
df
```

```python
df = pd.read_csv("students.csv", index_col=0)
df
```

> Load Excel files using pd.read_excel()
Pandas can also read Excel workbooks, commonly .xlsx files.

```python
df = pd.read_excel("students.xlsx") #to read an excel sheet
df = pd.read_excel("students.xlsx", sheet_name="Sheet1") #to read a specific excel sheet
df = pd.read_excel("students.xlsx", sheet_name=1) #to read a specific excel sheet using sheet number

excelFile = pd.ExcelFile("students.xlsx")
excelFile.sheet_names #to check the sheet names of the excel file
#If you get an error saying that an Excel-reading engine is missing, install openpyxl in your notebook's Python environment: pip install openpyxl

# For writing Excel files, Pandas also supports to_excel():
df.to_excel("output.xlsx", index=False)
```
