```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Employee": ["Ayaan", "Sara", "Rahul", "Sara", "Zoya",
                 "Ayaan", "Kabir", "Maya"],
    "Age": [25, np.nan, 30, np.nan, 28, 25,  np.nan, 35],
    "Salary": [40000, 50000, np.nan, 50000, 60000,
               40000, 45000, np.nan],
    "Department": ["IT", "HR", "IT", "HR", None, "IT", "Finance", "HR"]
})

df
```

```python
# 1 Missing-value analysis
# Beginner
# Using df:
# - Display the number of missing values in each column.
# - Calculate the percentage of missing values in each column, rounded to two decimal places.
# - Find the total number of missing cells in the entire DataFrame.

df.isna().sum()
df.isna().mean()*100
df.isna().sum().sum()
```

```python
# 2 Fill missing values intelligently
# Beginner–Intermediate
# Create a new DataFrame called df_filled in which:
# - Missing Age values are replaced with the median age.
# - Missing Salary values are replaced with the mean salary.
# - Missing Department values are replaced with "Unknown".
# - The original df must remain unchanged.
# Display df_filled.

df_filled = df.fillna({
    "Age":df["Age"].median(),
    "Salary":df["Salary"].mean(),
    "Department":"Unknown"
})

df_filled
```

```python
# 3 Drop rows using conditions
# Intermediate
# Create a new DataFrame called df_clean that:
# - Removes rows where Age is missing.
# - Also removes rows where Salary is missing.
# - Keeps the original DataFrame unchanged.
# - Resets the index so it starts from zero.
# Display df_clean.

df_clean = df.dropna(subset=["Age", "Salary"]).reset_index(drop=True)
df_clean
```

```python
# 4 Detect and remove duplicates
# Intermediate
# Using df:
# - Identify all rows that are duplicates based only on Employee and Salary.
# - Count how many rows are marked as duplicates using the default keep behavior.
# - Create df_unique by removing duplicates based on those two columns, keeping the last occurrence.

df.duplicated(["Employee","Salary"])
df.duplicated().sum()
df_unique = df.drop_duplicates(["Employee", "Salary"], keep="last" )
df_unique
```

```python
# 5 Complete data-cleaning challenge
# Intermediate
# Create a new DataFrame called df_final that:
# 1. Fills missing Age values with the median.
# 2. Fills missing Salary values with the median.
# 3. Fills missing Department values with "Unknown".
# 4. Removes duplicate records based on Employee and Salary, keeping the first occurrence.
# 5. Resets the index.
# 6. Displays the final DataFrame and its shape using .shape.
# Important: Perform the operations in the specified order.

df_final = df.fillna({
    "Age":df["Age"].median(),
    "Salary":df["Salary"].median(),
    "Department":"Unknown"
})
df_final = df_final.drop_duplicates(["Employee","Salary"])
df_final.reset_index(drop=True)
df_final.shape

```
